## Objective
Generate a sample dataset with Pandas, visualize the relationship between `f2` and `f3`, calculate the row-wise mean in `f5`, and export the completed DataFrame to Excel.

## 1. Import required libraries
Pandas is used for data handling, NumPy for reproducible random number generation, and Matplotlib for visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Create the DataFrame
The following cell generates 10 rows of random numerical values between 0 and 100 for columns `f1`, `f2`, `f3`, and `f4`. A fixed random seed makes the results reproducible.

```python
np.random.seed(42)
data = {
    'f1': np.random.uniform(0, 100, 10),
    'f2': np.random.uniform(0, 100, 10),
    'f3': np.random.uniform(0, 100, 10),
    'f4': np.random.uniform(0, 100, 10)
}

df = pd.DataFrame(data)
df = df.round(2)
df.head(10)
```

## 3. Scatter plot of f2 against f3
A scatter plot is used to examine how values in `f2` relate to values in `f3`. Each point represents one row in the DataFrame.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['f2'], df['f3'], alpha=0.8)
plt.xlabel('f2')
plt.ylabel('f3')
plt.title('Scatter Plot: f2 vs f3')
plt.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

## 4. Calculate the row-wise arithmetic mean
The mean is calculated across columns `f1`, `f2`, `f3`, and `f4` for each row. The result is stored in the new column `f5`.

```python
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1).round(2)
df
```

## 5. Export the DataFrame to Excel
The complete DataFrame, including columns `f1` through `f5`, is exported to an Excel file. The `index=False` argument prevents Pandas from adding an extra index column.

```python
output_file = 'Exercise1_Thongtsali_Sangtam.xlsx'
df.to_excel(output_file, index=False)
print(f'Excel file created successfully: {output_file}')
```

## 6. Conclusion
A DataFrame containing four randomly generated numerical columns was created. A scatter plot visualized the relationship between `f2` and `f3`. The row-wise mean was calculated and stored in `f5`, and the completed five-column DataFrame was exported to an Excel workbook.
